```python
# class Person:
#     name = ""
#     age = 0
#     address = ""

class Person:       # first letter of the class always has to be capital
    def __init__(self, name, age, address):     # self is the reference of the object in the memory of the system
        self.name = name
        self.age = age
        self.address = address
    def say_hi(self):
        print(f"Hello {self.name}")

    def __str__(self):     # the method having double underscore is known as dunder or magic method which doesnot have to be called
        return self.name
        
```

```python
# p1 = Person()
# # p1.address = "Pokhara"
# # p1.age = 23
# # p1.name = "Kritishma"

p1 = Person("Kritishma", 23, "Pokhara")
p1.say_hi()

print(p1)
```
