```python
subjects = ["nepali", "english", "maths", "science", "social"]
students_df["total_marks"] = students_df[subjects].sum(axis=1)
students_df
```

```python
def get_remark(total_marks):
    if total_marks > 450:
        return "Excellent"
    elif total_marks > 400:
        return "Very Good"
    elif total_marks > 350:
        return "Good"
    else:
        return "Need Improvement"

students_df["remark"] = students_df["total_marks"].apply(get_remark)

students_df
```

```python
def percentage(total_marks):
    return f"{(total_marks / 500) * 100:.2f}%"

students_df["percentage"] = students_df["total_marks"].apply(percentage)

students_df
```
