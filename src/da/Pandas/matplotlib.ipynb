```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# simple line plot
A = np.arange(0, 5, 0.1)

# simple y=mx line
B = A**2

plt.plot(A, B)
plt.title("Simple Line Plot")

# colour, linestyle, marker, linewidth, markersize, alpha, label
plt.plot(A, B, color = 'pink', linestyle = '--', marker = 'o', linewidth = 2)
```

```python
# bargraph
grade = ['A', 'B', 'C', 'D', 'F']
count = [10, 15, 20, 5, 2]
plt.bar(grade, count, color='green', alpha=0.59)
plt.title('Bar Graph')
plt.xlabel('Grades')
plt.ylabel('Count')

plt.show()
```

```python
# Data for the pie chart
labels = ['Apples', 'Bananas', 'Cherries', 'Dates']
sizes = [25, 30, 20, 25]  # Percentages or counts
colors = ['red', 'yellow', 'pink', 'orange']

 # Create the pie chart
plt.pie(sizes, labels=labels, colors=colors)
plt.title('Fruit Distribution', fontsize=14)

#legend
plt.legend(labels,title='Fruits',loc='center left',bbox_to_anchor=(1, 0.5))

plt.show()
```
