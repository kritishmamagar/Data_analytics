```python
arr = np.arange(1, 13).reshape(3, 4)
print("Array:\n", arr)
```

```python
print(arr[1:])
```

```python
print(arr[0:2, 1:3]) # here when it says 0:2 it only includes 0 and 1 row and the same goes for column too the ending number of the range is not included
```

```python
print(arr[1:, 1:3])
```

```python
print(arr[1:, ::2]) # Here the default value will be the starting and ending column number
```

```python
print(arr[1:, 1:3:2]) # Here we gave the range by putting 1 and 3
```

```python
print(arr[1:, 1::2]) # we can also give only starting or ending value
```

```python
data = np.array([12, 45, 7, 23, 56, 89, 3, 67])
iff = data > 10
print (data[iff])
```
