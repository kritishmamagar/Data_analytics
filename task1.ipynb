```python
## Task 1: Data Analysis with Python for students.csv file

import csv
file = 'students.csv'
with open(file, 'r') as f:
    reader = csv.DictReader(f)
    next(reader)
    print(f"{'Student ID':<10} {'Name':<20} {'Age':<5} {'Grade':<5}")
    for row in reader:
        print(f"{row['student_id']:<10} {row['name']:<20} {row['age']:<5} {row['grade']:<5}")
```

```python
## Task 1: Obtaining student's percentage

import csv
file = 'students.csv'
data = []
with open(file, 'r') as f:
    reader = csv.DictReader(f)
    next(reader)
    print(f"{'Student ID':<10} {'Name':<20} {'Age':<5} {'Grade':<5} {'Total Mark':<10} {'Average Mark':<12} {'Percentage':<10}")
    for row in reader:
        if row['grade'] == '10':
            total = int(row['nepali']) + int(row['english']) + int(row['maths']) + int(row['science']) + int(row['social'])
            average= total/5
            percentage = (total/500)*100
            print(f"{row['student_id']:<10} {row['name']:<20} {row['age']:<5} {row['grade']:<5} {total:<10} {average:<12} {percentage:<4.2f}%")

            data.append({
                'student_id': row['student_id'],
                'name': row['name'],
                'age': row['age'],
                'grade': row['grade'],
                'total_mark': total,
                'average_mark': average,
                'percentage': f'{percentage:.2f}%'           
            })
```

```python
filename = 'class10.csv'
with open(filename, 'w', newline='') as f:
          fieldnames = ['student_id', 'name', 'age', 'grade', 'total_mark', 'average_mark', 'percentage']
          writer = csv.DictWriter(f, fieldnames=fieldnames)
          writer.writeheader()
          writer.writerows(data)
```
